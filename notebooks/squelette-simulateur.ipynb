## L'exemple : un cinéma de plein air (fictif)

Un cinéma de plein air propose des soirées d'été. Chaque soir, il pleut avec une probabilité p, estimée avec son intervalle de confiance à 95 %. Deux options :

- **en plein air** : sans frais, mais une soirée sous la pluie est perdue ;
- **en salle** : une salle louée chaque soir, et aucune soirée perdue.

Laquelle rapporte le plus sur une saison ? Tous les chiffres sont fictifs.

## 1. Les paramètres

```python
# Les paramètres : le client les fait varier.
nb_soirees = 40          # soirées de la saison
p_pluie = 0.25           # probabilité qu'il pleuve un soir donné (estimation)
p_pluie_basse = 0.18     # borne basse de son intervalle de confiance à 95 %
p_pluie_haute = 0.32     # borne haute
recette_soiree = 900     # recette d'une soirée qui a lieu, en euros
location_salle = 250     # location de la salle, par soirée, en euros

# Les réglages de la simulation.
nb_saisons = 10_000      # saisons tirées au hasard
graine = 2026            # la même graine donne les mêmes résultats
```

## 2. La simulation

```python
import numpy as np
import matplotlib.pyplot as plt


def simuler(p, graine):
    """La recette de chaque saison simulée, pour chacune des deux options, en euros."""
    rng = np.random.default_rng(graine)
    soirs_de_pluie = rng.binomial(nb_soirees, p, size=nb_saisons)
    plein_air = (nb_soirees - soirs_de_pluie) * recette_soiree
    en_salle = np.full(nb_saisons, nb_soirees * (recette_soiree - location_salle))
    return plein_air, en_salle


def euros(x):
    """27 000 € plutôt que 27,000."""
    return f"{x:,.0f}".replace(",", "\u00a0") + "\u00a0€"


def pour_cent(x):
    """92 % plutôt que 0.92."""
    return f"{100 * x:.0f}\u00a0%"


plein_air, en_salle = simuler(p_pluie, graine)
```

## 3. La sortie : ce qui aide à décider

```python
print(f"En plein air : {euros(plein_air.mean())} par saison en moyenne ; "
      f"9 saisons sur 10 entre {euros(np.percentile(plein_air, 5))} "
      f"et {euros(np.percentile(plein_air, 95))}.")
print(f"En salle : {euros(en_salle.mean())} par saison, sans incertitude.")
print(f"Le plein air rapporte plus que la salle dans "
      f"{pour_cent(np.mean(plein_air > en_salle))} des saisons simulées.")

fig, ax = plt.subplots(figsize=(7, 3.5))
ax.hist(plein_air, bins=30, color="0.6", edgecolor="white", label="En plein air")
ax.axvline(en_salle[0], color="black", linestyle="--", label="En salle")
ax.set_xlabel("Recette de la saison (euros)")
ax.set_ylabel("Saisons simulées")
ax.set_title(f"{nb_saisons:,} saisons simulées, p = {p_pluie}".replace(",", "\u00a0"))
ax.legend()
plt.show()
```

## 4. Décider avec l'intervalle

La probabilité p est une estimation : la décision doit tenir de la borne basse à la borne haute de son intervalle (polycopié, chapitre 6, section 2.4). Si elle change en route, dites-le.

```python
for p in (p_pluie_basse, p_pluie, p_pluie_haute):
    plein_air_p, en_salle_p = simuler(p, graine)
    print(f"p = {p:.2f} : en plein air {euros(plein_air_p.mean())} en moyenne, "
          f"en salle {euros(en_salle_p.mean())} ; le plein air rapporte plus "
          f"dans {pour_cent(np.mean(plein_air_p > en_salle_p))} des saisons.")
```
